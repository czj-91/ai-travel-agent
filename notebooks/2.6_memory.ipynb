# 添加记忆

添加会话记忆（短期记忆）分为三步：

- 导入并初始化Checkpointer
- 创建Agent,指定Checkpointer
- 调用Agent,指定thread_id

````python
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver

agent = create_agent(
    "deepseek-chat",
    checkpointer=InMemorySaver()
)
````

````python
from langchain.messages import HumanMessage
from langchain_core.runnables import RunnableConfig

# 设定thread_id, 作为会话标识
config: RunnableConfig = {"configurable": {"thread_id": "thread_1"}}

# 第一次调用，告知AI我的信息
response = agent.invoke(
    {"messages": [HumanMessage(content="你好，我是Michale,我最喜欢猫猫。")]},
    config # 调用时添加thread_id
)
print(response)
````

````python
# 第二次调用，询问我的信息，这次带上thread_id,唤起记忆
from langchain.messages import HumanMessage

response = agent.invoke(
    {"messages": [HumanMessage(content="我最喜欢的动物是什么？")]},
    config # 调用时添加thread_id
)
print(response)
````

# Memory持久化存储

这里我们选择使用Sqlite作为存储方案，首先需要按照langgraph-checkpoint-sqlite依赖：

```bash
uv add langgraph-checkpoint-sqlite
```

接着，按照以下步骤使用：

- 导入依赖
- 初始化checkpoint
- 自动建表
- 创建Agent,指定checkpointer

````python
import sqlite3
from langgraph.checkpoint.sqlite import SqliteSaver
from pathlib import Path

# 目录不存在就自动创建：sqlite3.connect 只会创建 db 文件，不会创建父目录
Path("resources").mkdir(parents=True, exist_ok=True)


# 连接sqlite
connection = sqlite3.connect("resources/checkpoint.db", check_same_thread=False)
# 初始化checkpointer
checkpointer = SqliteSaver(connection)
# 自动建表
checkpointer.setup()

# 创建agent
agent = create_agent(
    "deepseek-chat",
    checkpointer=checkpointer,
)
````

````python
from langchain.messages import HumanMessage
from langchain_core.runnables import RunnableConfig

# 指定thread_id,作为会话标识
config: RunnableConfig = {"configurable": {"thread_id": "thread_2"}}

# 第一次调用，告知AI我的信息
response = agent.invoke(
    {"messages": [HumanMessage(content="你好，我是Michale,我最喜欢猫猫")]},
    config # 调用时添加thread_id
)
print(response)
````

````python
# 第二次调用，询问我的信息，这次带上thread_id,唤起记忆
response = agent.invoke(
    {"messages": [HumanMessage(content="我最喜欢的动物是什么？")]},
    config # 调用时添加thread_id
)
print(response)
````

# 记忆管理

当会话历史过长时，可能会超过模型的上下文窗口限制，常见的解决方案有：

- 修剪消息
- 删除消息
- 总结消息摘要

这里我们演示总结消息摘要的方案

````python
from langchain.agents import create_agent
from langchain.agents.middleware import SummarizationMiddleware
from langgraph.checkpoint.memory import InMemorySaver
from langchain_core.runnables import RunnableConfig

# 初始化checkpointer
checkpointer = InMemorySaver()
# 初始化中间件
middleware = SummarizationMiddleware(
    model="deepseek-chat",
    trigger=("messages", 3), # 触发时机，当消息数超过3条时，进行总结
    keep=("messages",1) # 保留的会话数，超过2条
)
# 创建Agent
agent = create_agent(
    "deepseek-chat",
    middleware=[middleware],
    checkpointer=checkpointer,
)

config: RunnableConfig = {"configurable": {"thread_id": "thread_3"}}
# 制造长会话历史
agent.invoke({"messages": [HumanMessage(content="你好，我是Michale")]}, config)
agent.invoke({"messages": [HumanMessage(content="我最喜欢的运动是乒乓球")]}, config)
agent.invoke({"messages": [HumanMessage(content="我最喜欢的动物是猫猫")]},config)
# 测试效果
final_response = agent.invoke({"messages": [HumanMessage(content="你还记得我吗？")]}, config)
````
